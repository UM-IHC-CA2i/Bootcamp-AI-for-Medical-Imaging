## Part 1: Python Basics

Python is the most widely used language in AI research. Let's start with the fundamentals.

### Variables and Types

A **variable** stores a value. Python figures out the type automatically.

```python
# Strings (text)
name = 'Radiology'
print(name)
print(type(name))

# Integers (whole numbers)
num_studies = 60
print(num_studies)
print(type(num_studies))

# Floats (decimal numbers)
accuracy = 0.85
print(accuracy)
print(type(accuracy))

# Booleans (True/False)
ai_flagged = True
print(ai_flagged)
print(type(ai_flagged))
```

### Lists

A **list** holds multiple values in order. You access items by their position (starting at 0).

```python
modalities = ['CR', 'CT', 'MR', 'US']

print('All modalities:', modalities)
print('First modality:', modalities[0])
print('Last modality:', modalities[-1])
print('Number of modalities:', len(modalities))
```

### For Loops

A **for loop** repeats an action for each item in a list.

```python
modalities = ['CR', 'CT', 'MR', 'US']

for mod in modalities:
    print(f'{mod} is an imaging modality')

print()

# Counting with a loop
total = 0
numbers = [10, 20, 30]
for n in numbers:
    total = total + n
print(f'Sum of {numbers} = {total}')
```

### Functions

A **function** is a reusable block of code. You define it with `def` and call it by name.

```python
def greet(name):
    return f'Welcome to AI research, {name}!'

message = greet('Dr. Smith')
print(message)

# A function that does a calculation
def percentage(part, whole):
    return round((part / whole) * 100, 1)

print(f'42 out of 60 = {percentage(42, 60)}%')
```

### Conditionals (if/else)

**Conditionals** let your code make decisions.

```python
confidence = 0.85

if confidence >= 0.9:
    print('High confidence')
elif confidence >= 0.7:
    print('Moderate confidence')
else:
    print('Low confidence')
```

## Part 2: Loading and Exploring Data

Now let's use Python to work with real-format data. We will use **pandas**, a library for tabular data (like a spreadsheet in Python).

```python
import pandas as pd
import matplotlib.pyplot as plt

print('Libraries loaded successfully!')
```

### Load the Dataset

We have a dataset of 60 radiology studies. Each row records whether an AI tool flagged a finding and whether a radiologist confirmed it.

**Important:** Replace `YOUR-USERNAME` below with your actual GitHub username.

### Explore the Data

Useful pandas commands:
- `df.head()` — shows the first 5 rows
- `df.shape` — gives you (rows, columns)
- `df['column'].value_counts()` — counts how many times each value appears

```python
# See the first 5 rows
df.head()
```

```python
# How big is the dataset?
print(f'Shape: {df.shape}')
print(f'This means {df.shape[0]} rows and {df.shape[1]} columns')
```

```python
# Count studies by imaging modality
print('Studies by modality:')
print(df['modality'].value_counts())
```

```python
# Count studies by body region
print('Studies by body region:')
print(df['body_region'].value_counts())
```

### Create a Bar Chart

Let's visualize how many studies used each imaging modality.

```python
counts = df['modality'].value_counts()

fig, ax = plt.subplots(figsize=(8, 5))
counts.plot(kind='bar', ax=ax, color='#0D9488')
ax.set_xlabel('Imaging Modality')
ax.set_ylabel('Number of Studies')
ax.set_title('Studies by Imaging Modality')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
